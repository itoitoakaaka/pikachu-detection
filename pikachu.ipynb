```python
import zipfile
import os

# トレーニング結果の重みファイルが保存されているディレクトリのパス
weights_dir = '/content/Pikachu-1/runs/detect/train3/weights/'
# 出力するZIPファイルのパス
zip_file_path = '/content/Pikachu-1/weights.zip'

# ZIPファイルを作成して圧縮する
with zipfile.ZipFile(zip_file_path, 'w') as zipf:
    # ディレクトリ内のファイルを追加する
    for root, dirs, files in os.walk(weights_dir):
        for file in files:
            file_path = os.path.join(root, file)
            zipf.write(file_path, arcname=os.path.relpath(file_path, weights_dir))

print(f'ZIP file created at {zip_file_path}')
```
